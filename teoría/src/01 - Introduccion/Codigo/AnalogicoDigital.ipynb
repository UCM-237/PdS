```python
import numpy as np
import matplotlib
from matplotlib import pyplot as plt
#%matplotib inline

def f_t(C: float, r: float, f: float, theta: float):
  return lambda t: np.real(C*np.exp(r*t) * np.exp(1j*(2*np.pi*f*t + theta)))

t = np.linspace(0, 10, num=1000, endpoint=True)
n = np.arange(0, 11)
x_range = n
y_range = np.arange(-2, 5)
f = 1/6
C = 5
r = -0.3
theta = -np.pi/2

x = f_t(C, r, f, theta)
x1 = x(t)
x2 = x(n)
x3 = x(t)
x4 = x(n)

plt.figure(dpi=360)
plt.subplot(2, 2, 1)
plt.grid(alpha=0.3)
plt.plot(t, x1)
plt.xticks(x_range)
plt.yticks(y_range)
plt.xlim(0, 10)
plt.title('Analógica')

plt.subplot(2, 2, 2)
plt.grid(alpha=0.3)
plt.stem(n, x2, basefmt='C0', markerfmt='C0.')
plt.xticks(x_range)
plt.yticks(y_range)
plt.xlim(0, 10)
plt.title('Muestreada')

plt.subplot(2, 2, 3)
plt.grid(alpha=0.3)
plt.plot(t, np.round(x3))
plt.xticks(x_range)
plt.yticks(y_range)
plt.xlim(0, 10)
plt.title('Cuantizada')

plt.subplot(2, 2, 4)
plt.grid(alpha=0.3)
plt.stem(n, np.round(x4), basefmt='C0', markerfmt='C0.')
plt.xticks(x_range)
plt.yticks(y_range)
plt.title('Digital')
plt.xlim(0, 10)


plt.tight_layout()
```
