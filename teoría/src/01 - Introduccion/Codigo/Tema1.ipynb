```python
import numpy as np
from bokeh.layouts import column, row
from bokeh.models import CustomJS, Slider
from bokeh.plotting import ColumnDataSource, figure, output_notebook, output_file, show
output_notebook()

t = np.linspace(0, 5, num=100)
f = 1
C = 1
r = 0.8
theta = 0
x = C*np.exp(r*t) * np.exp(1j*2*np.pi*f*t+theta)

source = ColumnDataSource(data=dict(t=t,x=np.real(x),x_t=np.abs(x), x_b=-np.abs(x)))

p = figure(width=300, height=300, tools=['hover','save'], x_axis_label='Tiempo (t)', y_axis_label='x(t)', title='Exponencial Continua')
p.xaxis.axis_label_text_font_size = '14px';
p.yaxis.axis_label_text_font_size = '14px';
p.line(x='t', y='x', source=source, line_color='red', line_width=2)
p.line(x='t', y='x_t', source=source, line_color='blue', line_width=1)
p.line(x='t', y='x_b', source=source, line_color='blue', line_width=1)

amp_slider = Slider(start=-1, end=1, value=1, step=.1, title="Amplitud (r)")
freq_slider = Slider(start=0.1, end=4, value=1, step=.1, title="Frecuencia (f)")
phase_slider = Slider(start=0, end=2*np.pi, value=0, step=.1, title="Fase (phi)")

callback = CustomJS(args=dict(source=source, amp=amp_slider, freq=freq_slider, phase=phase_slider),
                    code="""
    const data = source.data;
    const r = amp.value;
    const f = freq.value;
    const theta = phase.value;
    const t = data['t'];
    const x = data['x'];
    const x_t = data['x_t'];
    const x_b = data['x_b'];
    for (var i = 0; i < t.length; i++) {
        x[i] = Math.exp(r*t[i])*Math.cos(2*Math.PI*f*t[i]+theta);
        x_t[i] = +Math.abs(Math.exp(r*t[i]));
        x_b[i] = -Math.abs(x_t[i]);
    }
    source.change.emit();
""")
amp_slider.js_on_change('value', callback)
freq_slider.js_on_change('value', callback)
phase_slider.js_on_change('value', callback)

show(row(p, column(amp_slider, freq_slider, phase_slider)))

output_file('ExponencialComplejaContinua')
```

$x(t)=|C|e^{r t}e^{~j\omega t+\theta}$, donde $0\leq\omega=2\pi f\leq \infty$

Casos:
- $r<0$: Sinusoidal multiplicada por Exponencial decreciente
- $r=0$: Sinusoidal
- $r>0$: Sinusoidal multiplicada por Exponencial creciente

```python
import numpy as np
from bokeh.layouts import column, row
from bokeh.models import CustomJS, Slider
from bokeh.plotting import ColumnDataSource, figure, output_notebook, output_file, show
output_notebook()

# Exponencial continua
Tend = 1
t = np.linspace(0, Tend, num=200)
f = 1
xc = np.exp(1j*2*np.pi*f*t)

source1 = ColumnDataSource(data=dict(t=t, x1_r=np.real(xc), x1_i=np.imag(xc), x2_r=np.real(xc), x2_i=np.imag(xc), x1_m=np.abs(xc), x1_a=np.angle(xc), x2_m=np.abs(xc), x2_a=np.angle(xc)))

# Exponencial discreta
fs = 5
n = np.arange(0, Tend*fs)
xd = np.exp(1j*2*np.pi*(f/fs)*n)

source2 = ColumnDataSource(data=dict(t=n/fs, n=n, x_r=np.real(xd), x_i=np.imag(xd), x_m=np.abs(xd), x_a=np.angle(xd)))

p1 = figure(width=600, height=300, y_range=(-1.05, 1.05), tools=['hover','save'], title='Parte real de x(t)', x_axis_label='Tiempo (t)', y_axis_label='x(t)')
p1.xaxis.axis_label_text_font_size = '14px'
p1.yaxis.axis_label_text_font_size = '14px'
p1.line(x='t', y='x1_r', source=source1, line_color='red', line_dash='dotted', line_width=2)
p1.line(x='t', y='x2_r', source=source1, line_color='black', line_dash='dotted', line_width=1)
p1.circle(x='t', y='x_r', source=source2, line_color='red', line_width=4)

p2 = figure(width=600, height=300, y_range=(-1.05, 1.05), tools=['hover','save'], title='Parte imaginaria de x(t)', x_axis_label='Tiempo (t)', y_axis_label='x(t)')
p2.xaxis.axis_label_text_font_size = '14px'
p2.yaxis.axis_label_text_font_size = '14px'
p2.line(x='t', y='x1_i', source=source1, line_color='red', line_dash='dotted', line_width=2)
p2.line(x='t', y='x2_i', source=source1, line_color='black', line_dash='dotted', line_width=1)
p2.circle(x='t', y='x_i', source=source2, line_color='red', line_width=4)

p3 = figure(width=600, height=300, y_range=(0, 1.1), tools=['hover','save'], title='Módulo x(t)', x_axis_label='Tiempo (t)', y_axis_label='x(t)')
p3.xaxis.axis_label_text_font_size = '14px'
p3.yaxis.axis_label_text_font_size = '14px'
p3.line(x='t', y='x1_m', source=source1, line_color='red', line_dash='dotted', line_width=2)
p3.line(x='t', y='x2_m', source=source1, line_color='black', line_dash='dotted', line_width=1)
p3.circle(x='t', y='x_m', source=source2, line_color='red', line_width=4)

p4 = figure(width=600, height=300, y_range=(-np.pi, np.pi), tools=['hover','save'], title='Argumento x(t)', x_axis_label='Tiempo (t)', y_axis_label='x(t)')
p4.xaxis.axis_label_text_font_size = '14px'
p4.yaxis.axis_label_text_font_size = '14px'
p4.line(x='t', y='x1_a', source=source1, line_color='red', line_dash='dotted', line_width=2)
p4.line(x='t', y='x2_a', source=source1, line_color='black', line_dash='dotted', line_width=1)
p4.circle(x='t', y='x_a', source=source2, line_color='red', line_width=4)


freq_slider = Slider(start=1, end=10, value=1, step=1, title="Frecuencia (f)")
phase_slider = Slider(start=0, end=2*np.pi, value=0.1, step=1, title="Fase (phi)")

callback = CustomJS(args=dict(source1=source1, source2=source2, freq=freq_slider, phase=phase_slider, fs=fs),
                    code="""
    const data1 = source1.data;
    const data2 = source2.data;
    const f = freq.value;
    const theta = phase.value;
    const t = data1['t'];
    const x1_r = data1['x1_r'];
    const x1_i = data1['x1_i'];
    const x1a_r = data1['x2_r'];
    const x1a_i = data1['x2_i'];
    const x1_m = data1['x1_m'];
    const x1_a = data1['x1_a'];
    const x1a_m = data1['x2_m'];
    const x1a_a = data1['x2_a'];
    const t2 = data2['t'];
    const n = data2['n'];
    const x2_r = data2['x_r'];
    const x2_i = data2['x_i'];
    const x2_a = data2['x_a'];
    const fm = (f % fs);
    for (var i = 0; i < t.length; i++) {
        x1_r[i] = Math.cos(2*Math.PI*fm*t[i]+theta);
        x1_i[i] = Math.sin(2*Math.PI*fm*t[i]+theta);
        x1a_r[i] = Math.cos(2*Math.PI*f*t[i]+theta);
        x1a_i[i] = Math.sin(2*Math.PI*f*t[i]+theta);
        x1_a[i] = (2*Math.PI*fm*t[i]+theta + Math.PI) % 2*Math.PI - Math.PI;
        x1a_a[i] = (2*Math.PI*f*t[i]+theta + Math.PI) % 2*Math.PI - Math.PI;
    }
    for (var i = 0; i < n.length; i++) {
        t2[i] = n[i]/fs;
        x2_r[i] = Math.cos(2*Math.PI*(f/fs)*n[i]+theta);
        x2_i[i] = Math.sin(2*Math.PI*(f/fs)*n[i]+theta);
        x2_a[i] = (2*Math.PI*(f/fs)*n[i]+theta + Math.PI) % 2*Math.PI - Math.PI;
    }
    source1.change.emit();
    source2.change.emit();
""")
freq_slider.js_on_change('value', callback)
phase_slider.js_on_change('value', callback)

show(column(row(p1, p3), row(p2, p4), freq_slider, phase_slider))

output_file('Exponencial Continua vs. Discreta')
```

```python
import numpy as np
from bokeh.layouts import column, row
from bokeh.models import CustomJS, Slider
from bokeh.plotting import ColumnDataSource, figure, output_notebook, show
output_notebook()

f1 = 1        # Frecuencia fundamental
fs = 4*f1     # Frecuencia de muestreo
f2 = f1 + fs  # Frecuencia harmónica
Tend = 2 / f1 # periodos

N = round(Tend * fs) + 1 # Muestras
M = 50                   # Sobremuestreo

tc = np.linspace(0, Tend, num=N*M)
td = np.linspace(0, Tend, num=N)
n = np.arange(0, N)
x1c = np.cos(2*np.pi*f1*tc)
x2c = np.cos(2*np.pi*f2*tc)
x1d = np.cos(2*np.pi*(f1/fs)*n)
x2d = np.cos(2*np.pi*(f2/fs)*n)

src_c = ColumnDataSource(data=dict(tc=tc, x1c=x1c, x2c=x2c))
src_d = ColumnDataSource(data=dict(td=td, x1d=x1d, x2d=x2d))

p1 = figure(width=700, height=300, tools=['hover','save'], title='Parte real de x(t)', x_axis_label='Tiempo (t)', y_axis_label='x(t)')
p1.xaxis.axis_label_text_font_size = '14px';
p1.yaxis.axis_label_text_font_size = '14px';
p1.line(x='tc', y='x1c', source=src_c, line_color='red', line_width=2)
p1.line(x='tc', y='x2c', source=src_c, line_color='green', line_width=2)
p1.circle(x='td', y='x1d', source=src_d, line_color='blue', line_width=2)
p1.square(x='td', y='x2d', source=src_d, line_color='blue', line_width=2)

freq_slider = Slider(start=-4, end=4, value=1, step=1, title="Frecuencia (f)")
#phase_slider = Slider(start=0, end=2*np.pi, value=0, step=.1, title="Fase ($\\theta$)")
print(f1)
print(fs)

callback = CustomJS(args=dict(source=src_c, freq=freq_slider, params=[f1, fs]),
                    code="""
    const data = source.data;
    const k = freq.value;
    const f1 = params[0];
    const fs = params[1];
    const t = data['tc'];
    const x2c = data['x2c'];
    for (var i = 0; i < t.length; i++) {
        x2c[i] = Math.cos(2*Math.PI*(f1 + k*fs/2)*t[i]);
    }
    source.change.emit();
""")
freq_slider.js_on_change('value', callback)

show(column(p1, freq_slider))
```
